->**Empty Cells**

Empty cells can potentially give you a wrong result when you analyze data.



Remove Rows:-

The result from the converting in the example above gave us a NaT value, which can be handled as a NULL value, and we can remove the row by using the dropna() method.


eg :-Return a new Data Frame with no empty cell

```python
import pandas as pd
df=pd.read_excel('/content/drive/MyDrive/Colab Notebooks/dataset/workout_data.xlsx')
new_df=df.dropna()
print(new_df.to_string())
```

 the dropna() method returns a new DataFrame, and will not change the original.

If you want to change the original DataFrame, use the inplace = True argument:

eg :-Remove all rows with NULL values:

```python
import pandas as pd
df=pd.read_excel('/content/drive/MyDrive/Colab Notebooks/dataset/workout_data.xlsx')
df.dropna(inplace=True)
print(df.to_string)
```

the dropna(inplace = True) will NOT return a new DataFrame, but it will remove all rows containing NULL values from the original DataFrame.

**Replace Empty Values**

Another way of dealing with empty cells is to insert a new value instead.


This way you do not have to delete entire rows just because of some empty cells.


The fillna() method allows us to replace empty cells with a value:

 eg:-Replace NULL values with the number 130:

```python
import pandas as pd
df=pd.read_excel('/content/drive/MyDrive/Colab Notebooks/dataset/workout_data.xlsx')
df.fillna(130,inplace=True)
print(df.to_string())
```

**Replace Only For Specified Columns**

The example above replaces all empty cells in the whole Data Frame.


To only replace empty values for one column, specify the column name for the DataFrame:

eg:-Replace NULL values in the "Calories" columns with the number 130:

```python
import pandas as pd
df=pd.read_excel('/content/drive/MyDrive/Colab Notebooks/dataset/workout_data.xlsx')
df.fillna({'Calories':130},inplace=True)
print(df.to_string())
```

**Replace Using Mean, Median, or Mode**

A common way to replace empty cells, is to calculate the mean, median or mode value of the column.


Pandas uses the mean() median() and mode() methods to calculate the respective values for a specified column:

EG:-Calculate the MEAN, and replace any empty values with it:

eg:-Calculate the MEAN, and replace any empty values with it:

```python
import pandas as pd
df=pd.read_excel('/content/drive/MyDrive/Colab Notebooks/dataset/workout_data.xlsx')
x=df['Calories'].mean()
df.fillna({'Calories':x},inplace=True)
print(df.to_string())
```

eg:-Calculate the MEDIAN, and replace any empty values with it:

```python
import pandas as pd
df=pd.read_excel('/content/drive/MyDrive/Colab Notebooks/dataset/workout_data.xlsx')
x=df['Calories'].median()
df.fillna({'Calories':x},inplace=True)
print(df.to_string())
```

eg:-Calculate the MODE, and replace any empty values with it:

```python
import pandas as pd
df=pd.read_excel('/content/drive/MyDrive/Colab Notebooks/dataset/workout_data.xlsx')
x=df['Calories'].mode()[0]
df.fillna({'Calories':x},inplace=True)
print(df.to_string())
```

**Pandas - Fixing Wrong Data**

---



**Wrong Data**

"Wrong data" does not have to be "empty cells" or "wrong format", it can just be wrong, like if someone registered "199" instead of "1.99".

If you take a look at our data set, you can see that in row 7, the duration is 450, but for all the other rows the duration is between 30 and 60.

One way to fix wrong values is to replace them with something else.

In our example, it is most likely a typo, and the value should be "45" instead of "450", and we could just insert "45" in row 7:

```python
import pandas as pd
df=pd.read_excel('/content/drive/MyDrive/Colab Notebooks/dataset/workout_sample.xlsx')
df.loc[7,'Duration']=45
print(df.to_string())
```

```python
df.dropna(inplace=True)
print(df.to_string())
```

For small data sets you might be able to replace the wrong data one by one, but not for big data sets.

To replace wrong data for larger data sets you can create some rules, e.g. set some boundaries for legal values, and replace any values that are outside of the boundaries.

eg:-Loop through all values in the "Duration" column.

If the value is higher than 120, set it to 120:

```python
for x in df.index:
  if df.loc[x,'Duration']>120:
    df.loc[x,'Duration']=120
```

Another way of handling wrong data is to remove the rows that contains wrong data.

eg :Delete rows where "Duration" is higher than 120

```python
for x in df.index:
  if df.loc[x,"Duration"]>120:
    df.drop(x,inplace=True)
```

##Concat
It is used to concat two or more dataframes

```python
import pandas as pd
tcs={'id':[101,102,103,104],'name':['mithun','dipin','jose','rahul'],
     'age':[25,26,27,29]}

wipro={'id':[101,102,103,104],
       'name':['dev','sreenath','sreerag','shafeel'],
       'age':[24,24,27,26]}

tcs_emp=pd.DataFrame(tcs)
wipro_emp=pd.DataFrame(wipro)
tcs_emp
```

```python
wipro_emp
```

```python
pd.concat([tcs_emp,wipro_emp])
```

ignore index-To get common index after concatenation

```python
pd.concat([tcs_emp,wipro_emp],ignore_index=True)
```

keys-To provide seperate keys for each concatenate data frames

```python
data=pd.concat([tcs_emp,wipro_emp],keys=['tcs','wipro'])
data
```

concatenating vertically

```python
data=pd.concat([tcs_emp,wipro_emp],axis=1)
data
```

concatinating a dataframe with a series

```python
salary=pd.Series([65000,75000,15000,25000],name='salary')
pd.concat([tcs_emp,salary],axis=1)
```

##Merge
Merge is a process to merge two tables horizontally

```python
import pandas as pd
tcs={'id':[101,102,103,106],'name':['mithun','dipin','joe','rahul'],
     'age':[25,26,27,29]}

wipro={'id':[101,102,103,105],
       'name':['dev','sreenath','sreerag','shafeel'],
       'age':[24,24,27,26]}

tcs_emp=pd.DataFrame(tcs)
wipro_emp=pd.DataFrame(wipro)

tcs_emp=pd.DataFrame(tcs)
tcs_emp
```

```python
wipro_emp=pd.DataFrame(wipro)
wipro_emp
```

```python
pd.merge(tcs_emp,wipro_emp,on='id')
```

only  respective number will appear if on='id' given

suffixes

```python
pd.merge(tcs_emp,wipro_emp,on='id', suffixes=['left','right'])
```

outer join

```python
pd.merge(tcs_emp,wipro_emp,on='id',how='outer')
```

```python
pd.merge(tcs_emp,wipro_emp,on='id',how='outer',indicator=True)
```

left join

```python
pd.merge(tcs_emp,wipro_emp,on='id',how='left')
```

right join

```python
pd.merge(tcs_emp,wipro_emp,on='id',how='right')
```

##Cross tab or contingency table

In pandas, pd.crosstab() is used to compute a cross-tabulation of two or more factors. It’s often used for analyzing categorical variables, similar to a contingency table in statistics.

syntax:

pd.crosstab(index, columns, values=None, aggfunc=None, margins=False, margins_name='All')


```python
import pandas as pd
df=pd.read_csv('/content/drive/MyDrive/Colab Notebooks/dataset/survey_crosstab.csv')
df
```

create cross_tabulation

```python
ct=pd.crosstab(df['Gender'],df['Preferred_Device'])
print(ct)
```

Cross-tab of AgeGroup vs Preferred_Device with totals:

```python
ct2=pd.crosstab(df['AgeGroup'],df['Preferred_Device'],margins=True,margins_name='Total')
print(ct2)
```
